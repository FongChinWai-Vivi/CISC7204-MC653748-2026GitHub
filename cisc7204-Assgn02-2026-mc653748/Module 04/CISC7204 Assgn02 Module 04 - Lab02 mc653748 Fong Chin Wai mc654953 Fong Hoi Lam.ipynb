## Goal

Create a dashboard that displays the percentage of flights running under a specific distance group. Distance group is the distance interval, every 250 miles, for a flight segment.

## Task 1 – Data Preparation

Import the required libraries, read the dataset, sample 500 random data points, and create the pie chart.

````python
# Import required packages
import pandas as pd
import plotly.express as px
import dash
from dash import dcc
from dash import html

# Read the airline data into pandas dataframe (local file)
airline_data = pd.read_csv('airline_data.csv',
                           encoding = "ISO-8859-1",
                           dtype={'Div1Airport': str, 'Div1TailNum': str,
                                  'Div2Airport': str, 'Div2TailNum': str})
print('Data loaded. Shape:', airline_data.shape)
````

````python
# Randomly sample 500 data points. Setting the random state to be 42 so that we get same result.
data = airline_data.sample(n=500, random_state=42)
data.shape
````

````python
# Pie Chart Creation
fig = px.pie(data, values='Flights', names='DistanceGroup', title='Distance group proportion by flights')
fig.show()
````

## Task 2 – Create the Dash application and layout skeleton

Mapping of dashboard components to Dash HTML tags:
- Title: `html.H1()`
- Description: `html.P()`
- Chart: `dcc.Graph()`

````python
# Create a dash application
app = dash.Dash(__name__)

# Get the layout of the application and adjust it.
# Create an outer division using html.Div and add title to the dashboard using html.H1 component
# Add description about the graph using HTML P (paragraph) component
# Finally, add graph component.
app.layout = html.Div(children=[html.H1('Airline Dashboard', style={'textAlign': 'center', 'color': '#503D36', 'font-size': 40}),
                                html.P('Proportion of distance group (250 mile distance interval group) by flights.', style={'textAlign':'center', 'color': '#F57241'}),
                                dcc.Graph(figure=fig),
                    ])
print('Dash app layout created.')
````

## Run the application

The application is designed to be run as a standalone script (e.g. `python dash_basics.py`) which starts a local Dash server. Inside this notebook the server is **not** started so that the kernel does not block; the code below is kept for reference.
